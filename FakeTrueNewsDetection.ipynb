```python
# 1. Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# 2. Load data (upload Fake.csv and True.csv in Colab)
fake = pd.read_csv("/content/Fake.csv")
real = pd.read_csv("/content/True.csv")

# 3. Add labels
fake["label"] = 0   # 0 = Fake
real["label"] = 1   # 1 = Real

# 4. Combine datasets
data = pd.concat([fake, real]).reset_index(drop=True)

# 5.Clean data
data.isnull().sum()
data.dropna(inplace=True)
data.drop_duplicates(inplace=True)

# 6.Print columns to debug
print("Columns after concatenation:", data.columns)

# 7. Keep only the text and label
data = data[["text", "label"]]

# 8. Quick data info
print("Dataset shape:", data.shape)
print(data.head())
print(data["label"].value_counts())


# 9. Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    data["text"], data["label"], test_size=0.2, random_state=42
)

# 10. TF-IDF vectorization
tfidf = TfidfVectorizer(stop_words="english", max_df=0.7)
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# 11. Train Logistic Regression model
model = LogisticRegression()
model.fit(X_train_tfidf, y_train)


# 12. Predictions and evaluation
y_pred = model.predict(X_test_tfidf)


print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# ✅ 13. Reward / Punishment System (Simulation)
reward = 0
for true_label, pred_label in zip(y_test, y_pred):
    if true_label == pred_label:
        reward += 1   # Correct → Reward
    else:
        reward -= 1   # Incorrect → Punishment

print("\nTotal Reward Score:", reward)
```
